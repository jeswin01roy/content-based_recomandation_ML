```python
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer

from sklearn.metrics.pairwise import cosine_similarity
```

```python
import pandas as pd

df = pd.DataFrame({
    "Product_ID": [201, 202, 203, 204, 205, 206, 207, 208, 209, 210],

    "Product_Name": [
        "Running Shoes",
        "Casual Shoes",
        "T-Shirt",
        "Jeans",
        "Backpack",
        "Wireless Mouse",
        "Keyboard",
        "Water Bottle",
        "Coffee Mug",
        "Bluetooth Speaker"
    ],

    "Category": [
        "Footwear",
        "Footwear",
        "Clothing",
        "Clothing",
        "Bags",
        "Electronics",
        "Electronics",
        "Accessories",
        "Kitchen",
        "Electronics"
    ],

    "Brand": [
        "Nike",
        "Adidas",
        "Puma",
        "Levi's",
        "Skybags",
        "Logitech",
        "HP",
        "Milton",
        "Cello",
        "JBL"
    ],

    "Price": [
        4999,
        4499,
        999,
        2499,
        1999,
        1499,
        1899,
        499,
        299,
        3499
    ],

    "Color": [
        "Black",
        "White",
        "Blue",
        "Navy",
        "Grey",
        "Black",
        "Black",
        "Green",
        "White",
        "Red"
    ],

    "Size": [
        "9",
        "8",
        "M",
        "32",
        "Large",
        "Standard",
        "Standard",
        "1 L",
        "350 ml",
        "Standard"
    ]
})

# Display the DataFrame
print(df)
```

```python
df["Features"] = (
      df["Product_Name"]
    + " "
    + df["Category"]
    + " "
    + df["Brand"]
    + " "
    + df["Color"]
)

print("\nCombined Features")
print("-"*50)
print(df[["Product_Name", "Features"]])
```

```python
cv = CountVectorizer()

feature_matrix = cv.fit_transform(df["Features"])

print("\nFeature Matrix Shape")
print(feature_matrix.shape)


print(feature_matrix)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

price_scaled = scaler.fit_transform(df[["Price"]])
```

```python
from scipy.sparse import csr_matrix

price_sparse = csr_matrix(price_scaled)
```

```python
print(price_sparse)
```

```python
from scipy.sparse import hstack

combined_matrix = hstack([feature_matrix, price_sparse])
```

```python
print(combined_matrix)
```

```python
similarity = cosine_similarity(feature_matrix)

print("\nSimilarity Matrix")
print("-"*50)
print(similarity)
```

```python
def recommendation_table():

    result = []
    for product_name in df["Product_Name"]:

    
        index = df[df["Product_Name"] == product_name].index[0]


        similarity_scores = list(enumerate(similarity[index]))


        filtered_products = [
            (i, score)
            for i, score in similarity_scores
            if score > 0 and i != index
        ]
      
        sorted_products = sorted(
            filtered_products,
            key=lambda x: x[1],
            reverse=True
        )

        recs = [
            f"{df.iloc[i]['Product_Name']} ({score:.3f})"
            for i, score in sorted_products[:5]
        ]

    
        result.append({
            "Product_Name": product_name,
            "Rec1": recs[0] if len(recs) > 0 else None,
            "Rec2": recs[1] if len(recs) > 1 else None,
            "Rec3": recs[2] if len(recs) > 2 else None,
            "Rec4": recs[3] if len(recs) > 3 else None,
            "Rec5": recs[4] if len(recs) > 4 else None
        })


    recommendation_df = pd.DataFrame(result)

    return recommendation_df
```

```python
recommendation_df = recommendation_table()

recommendation_df
```
